```python
import os

def check_classes_in_labels(labels_dirs, max_class_id):
    print("Проверка классов в аннотациях...")
    total_files = 0
    files_with_errors = 0
    classes_found = set()

    for labels_dir in labels_dirs:
        if not os.path.exists(labels_dir):
            print(f"Папка не найдена: {labels_dir}")
            continue
        for filename in os.listdir(labels_dir):
            if not filename.endswith(".txt"):
                continue
            total_files += 1
            filepath = os.path.join(labels_dir, filename)
            with open(filepath, 'r') as f:
                lines = f.readlines()
            for i, line in enumerate(lines):
                parts = line.strip().split()
                if len(parts) == 0:
                    continue
                class_id = parts[0]
                try:
                    class_id_int = int(class_id)
                except ValueError:
                    print(f"Ошибка в файле {filepath}, строка {i+1}: класс не число — '{class_id}'")
                    files_with_errors += 1
                    continue
                classes_found.add(class_id_int)
                if class_id_int < 0 or class_id_int > max_class_id:
                    print(f"Ошибка в файле {filepath}, строка {i+1}: класс {class_id_int} вне диапазона 0-{max_class_id}")
                    files_with_errors += 1

    print(f"\nВсего файлов аннотаций: {total_files}")
    print(f"Уникальные классы, найденные в аннотациях: {sorted(classes_found)}")
    if files_with_errors == 0:
        print("Ошибок классов не найдено.")
    else:
        print(f"Файлов с ошибками: {files_with_errors}")


if __name__ == "__main__":

    labels_dirs = [
        r"C:\Users\Legion\PycharmProjects\YOLOTrain\dataset\dronedetect\train\labels",
        r"C:\Users\Legion\PycharmProjects\YOLOTrain\dataset\dronedetect\valid\labels",
        r"C:\Users\Legion\PycharmProjects\YOLOTrain\dataset\dronedetect\test\labels"
    ]
    max_class_id = 0  # если 1 класс, то max_class_id=0
    check_classes_in_labels(labels_dirs, max_class_id)
```
